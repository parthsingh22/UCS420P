Question 1 — Lists

Take your roll number, extract its individual digits, and multiply each digit by
10 to form the list L.

```python
L = [d * 10 for d in digits]
L_original = L.copy()

print("L =", L)
print("Length of L:", len(L))
```

1(i) — Print L

```python
print("L =", L)
```

1(ii) — append() and insert()
append(x) always adds to the **end**; insert(i, x) puts the value **at index `i`**
and shifts everything from that position one place to the right.

```python
L.append(45)
print("After append(45) :", L)


L.insert(6, 60)
print("After insert(6,60):", L)
```

1(iii)-remove() and pop()
remove(value) deletes the **first occurrence of a value** and returns nothing.
pop(index) deletes **by position** and *returns* the removed element.

```python
L.remove(60)
print("After remove(60):", L)


popped = L.pop(2)
print("After pop(2)    :", L)
print("pop(2) returned :", popped)
```

1(iv)-Sorting

```python
L.sort()
print("Ascending  :", L)

L.sort(reverse=True)
print("Descending :", L)
```

1(v)-Slicing

```python
print("First three elements:", L[:3])
print("Last three elements :", L[-3:])
```

1(vi)-List comprehension: elements greater than the average

```python
average = sum(L) / len(L)
above_average = [x for x in L if x > average]

print("Average of L :", average)
print("Elements above avg :", above_average)
```

Question 3 — Random numbers

The seed is the roll number, so the list looks random but is **reproducible**
and different for every student.

3(i) — 100 random numbers between 100 and 900 (inclusive)

```python
random.seed(int(ROLL_NO))

numbers = [random.randint(100, 900) for _ in range(100)]

print("First 20 of the 100 numbers:", numbers[:20])
print("Total generated:", len(numbers))
```

3(ii) — Odd numbers

```python
odds = [n for n in numbers if n % 2 != 0]

print("Count of odd numbers:", len(odds))
print("Odd numbers:", odds)
```

3(iii) — Even numbers

```python
evens = [n for n in numbers if n % 2 == 0]

print("Count of even numbers:", len(evens))
print("Even numbers:", evens)
```

3(iv) — Prime numbers

```python
def is_prime(n):
    """True if n is a prime number."""
    if n < 2:
        return False
    if n % 2 == 0:
        return n == 2
    for factor in range(3, int(n ** 0.5) + 1, 2):
        if n % factor == 0:
            return False
    return True


primes = [n for n in numbers if is_prime(n)]

print("Count of prime numbers:", len(primes))
print("Prime numbers:", primes)
```

3(v) — Most frequent number

```python
counts = Counter(numbers)
number, frequency = counts.most_common(1)[0]

print(f"Most frequent number: {number} (occurs {frequency} time(s))")


top_all = [n for n, c in counts.items() if c == frequency]
print("All numbers sharing that frequency:", sorted(top_all))
```

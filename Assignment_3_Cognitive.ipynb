Q.6 Create a sample dataset (employees.csv) containing information about employees in
a company.

```python
import pandas as pd

data = {
    "Employee_ID": [101, 102, 103, 104, 105],
    "Name": ["Alice", "Bob", "Charlie", "Diana", "Edward"],
    "Department": ["HR", "IT", "IT", "Marketing", "Sales"],
    "Age": [29, 34, 41, 28, 38],
    "Salary": [50000, 70000, 65000, 55000, 60000],
    "Years_of_Experience": [4, 8, 10, 3, 12],
    "Joining_Date": ["2020-03-15", "2017-07-19", "2013-06-01", "2021-02-10", "2010-11-25"],
    "Gender": ["Female", "Male", "Male", "Female", "Male"],
    "Bonus": [5000, 7000, 6000, 4500, 5000],
    "Rating": [4.5, 4.0, 3.8, 4.7, 3.5]
}

df = pd.DataFrame(data)

df.to_csv("employees.csv", index=False)
df = pd.read_csv("employees.csv")

print("Shape:")
print(df.shape)

print("\nDataFrame Information:")
df.info()

print("\nDescriptive Statistics:")
print(df.describe())

print("\nFirst 5 rows:")
print(df.head())

print("\nLast 3 rows:")
print(df.tail(3))

print("\nAverage Salary:", df["Salary"].mean())
print("Total Bonus:", df["Bonus"].sum())
print("Youngest Age:", df["Age"].min())
print("Highest Rating:", df["Rating"].max())

df = df.sort_values("Salary", ascending=False)

print("\nSorted by Salary:")
print(df)

df["Performance"] = df["Rating"].apply(
    lambda x: "Excellent" if x >= 4.5 else "Good" if x >= 4.0 else "Average"
)

print("\nPerformance:")
print(df[["Name", "Rating", "Performance"]])

print("\nMissing Values:")
print(df.isnull().sum())

df.rename(columns={"Employee_ID": "ID"}, inplace=True)

print("\nEmployees with more than 5 years of experience:")
print(df[df["Years_of_Experience"] > 5])

print("\nEmployees in IT department:")
print(df[df["Department"] == "IT"])

df["Tax"] = df["Salary"] * 0.10

print("\nFinal DataFrame:")
print(df)

df.to_csv("employees_modified.csv", index=False)

print("\nModified dataset saved successfully.")
```
